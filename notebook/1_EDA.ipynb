```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("data/data.csv")
df.head()
```

```python
df_copy=df.copy(deep=True)
df_1=df.drop(["customerID"], axis=1)
df_1.head()
```

```python
df_1.info()
```

```python
for col in df_1.columns.tolist():
    print(col)
    print(df_1[col].dtype)
    print(df_1[col].unique())
    print("#----------------------")
```

```python
df_1["TotalCharges"]=pd.to_numeric(df_1["TotalCharges"],errors="coerce")
```

```python
df_1.duplicated().sum()
```

```python
df_1.isnull().sum()
```

```python
df_1.columns.tolist()
```

```python
cat_cols=df_1.select_dtypes(include="object").columns.tolist()
num_cols=df_1.select_dtypes(include="number").columns.tolist()
print(cat_cols)
print(num_cols)
```

```python
for col in num_cols:
    plt.figure(figsize=(8,5))
    sns.histplot(df_1[col], bins=100)
    plt.title(col)
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.show()
```

```python
for col in cat_cols:
    print(f"{col}")
    print(df_1[col].value_counts())
```

```python
cat_cols
```

```python
for cat in cat_cols:
    plt.figure(figsize=(4,2))
    sns.countplot(data=df_1, x=cat)
    plt.xlabel(f"{cat}")
    plt.ylabel("Counts")
    plt.title(f"{cat} bar plot")
    plt.xticks(rotation=45)
    plt.show()
```

```python
df_1["Churn_label"]=df_1["Churn"].map({
    "Yes":1, "No":0
})
df_1["Churn_label"]
```

```python
for cat in cat_cols:
    plt.figure(figsize=(4,2))
    sns.barplot(data=df_1,x=cat, y="Churn_label", estimator="mean",errorbar=None)
    plt.xlabel(cat)
    plt.ylabel("Percentage")
    plt.xticks(rotation=45)
    plt.title(f"{cat} churn")
    plt.show()
```

```python
df_1
```

```python
df_encoded=pd.get_dummies(
    df_1[cat_cols],
    drop_first=True, dtype=int
)
```

```python
df_encoded=pd.concat([df_encoded, df_1[num_cols]], axis=1)
```

```python
df_encoded.head()
```

```python
corr=df_encoded.corr()
plt.figure(figsize=(30,30))
sns.heatmap(corr, annot=True, cmap="coolwarm", center=0)
plt.show()
```
